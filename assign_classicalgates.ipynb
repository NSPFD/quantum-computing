## Question 1 (8 points)

Complete the function `OR3` that implements the 3-qubit to 1-qubit $OR$ operator. This operator acts on 4 qubits in the state $|q_0,q_1,q_2,q_3\rangle$ and maps it to $|q_0,q_1,q_2,q_3 \oplus (q_0 \text{ or } q_1 \text{ or } q_2)  \rangle$.

The function `OR3` has

- Inputs: None
- Returns: A `QuantumCircuit` object with 4 qubits and 1 classical bit that implements the $OR$ operator as described above.

Don't add any measurements to the returned circuit.

```python
from qiskit import QuantumCircuit

def OR3():
    # Initialize a circuit with 4 quantum qubits and 1 classical bit
    qc = QuantumCircuit(4, 1)

    # Inputs: q0, q1, q2 (indices 0, 1, 2)
    # Target: q3 (index 3)

    # YOUR CODE HERE
    # 1. Invert the input qubits so we can look for the "all zeros" condition
    qc.x([0, 1, 2])

    # 2. Invert the target qubit before checking, so we can flip it back
    # if the "all zeros" state is detected
    qc.x(3)

    # 3. Apply a Multi-Controlled X gate.
    # If q0, q1, and q2 were originally all 0, they are now all 1,
    # which triggers the MCX and flips the target (q3) back.
    qc.mcx([0, 1, 2], 3)

    # 4. Uncompute the X gates to restore the control qubits and target logic
    qc.x([0, 1, 2])

    return qc
```

```python
from qiskit import QuantumCircuit
from qiskit.circuit.library.standard_gates import XGate
from qiskit.primitives import StatevectorSampler

def OR3():
    circuit = QuantumCircuit(4)
    # Do not modify anything above this line

    # YOUR CODE HERE
    # 1. Invert the control qubits (0, 1, 2) to check for the "all zeros" state
    circuit.x([0, 1, 2])

    # 2. Set up the target qubit (3) negation
    circuit.x(3)

    # 3. Apply the Multi-Controlled X gate (TOFFOLI-3 variant)
    # This flips qubit 3 back if and only if qubits 0, 1, and 2 are all 1s (originally 0s)
    circuit.mcx([0, 1, 2], 3)

    # 4. Restore the control qubits back to their original input states
    circuit.x([0, 1, 2])

    # Do not modify anything below this line
    return circuit
```

```python
def test(input):
    circuit = QuantumCircuit(4, 1)

    # implement the input state with a suitable use of X gates.
    for i in range(3):
        if input[i]=='1':
            circuit.x(i)

    # Apply the OR3 gate you created
    circuit.compose(OR3(), range(4), inplace=True)

    circuit.measure(3, 0)

    sampler = StatevectorSampler()
    job = sampler.run([circuit], shots=100)
    result = job.result()[0]
    counts = result.data.c.get_counts()
    return counts

test('101')
```
